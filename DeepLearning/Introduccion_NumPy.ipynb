## Arrays

Un **array** es una estructura de datos que consiste en una colección de elementos (valores o variables), cada uno identificado por al menos un índice o clave. Un array se almacena de modo que la posición de cada elemento se pueda calcular a partir de su tupla de índice mediante una fórmula matemática. El tipo más simple de array es un array lineal, también llamado array unidimensional.

En numpy:
* Cada dimensión se denomina **axis**
* El número de dimensiones se denomina **rank**
* La lista de dimensiones con su correspondiente longitud se denomina **shape**
* El número total de elementos (multiplicación de la longitud de las dimensiones) se denomina **size**

```python
# Array cuyos valores son todos 0
a = np.zeros((2, 4)) # dos filas y cuatro columnas
print(a)
```

_**a**_ es un array:
* Con dos **axis**, el primero de longitud 2 y el segundo de longitud 4
* Con un **rank** igual a 2
* Con un **shape** igual (2, 4)
* Con un **size** igual a 8

```python
a.shape 
```

```python
# numero de dimensiones del array
a.ndim
```

```python
a.size 
```

## Acceso a los elementos de un array

### Array unidimensional

```python
# Creación de un Array unidimensional
array_uni = np.array([1, 3, 5, 7, 9, 11])
print("Shape:", array_uni.shape)
print("Array_uni:", array_uni)
```

```python
# Accediendo al quinto elemento del Array
array_uni[4]
```

```python
# Accediendo al tercer y cuarto elemento del Array
array_uni[2:4] # slaicing de elementos
```

```python
# Accediendo a los elementos 0, 3 y 5 del Array
array_uni[0::3] # desde el elemento 0 hasta el final, saltando de 3
```

### Array multidimensional

```python
# Creación de un Array multidimensional
array_multi = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print("Shape:", array_multi.shape)
print("Array_multi:\n", array_multi)
```

```python
# Accediendo al cuarto elemento del Array
print("Cuarto elemento:", (array_multi[0, 3])) # fila 0, columna 3
print("Sexto elemento:", (array_multi[1, 2])) # fila 1, columna 2
```

```python
# Accediendo a la segunda fila del Array
array_multi[1, :]
```

```python
# Accediendo al tercer elemento de las dos primeras filas del Array
array_multi[0:2, 2]
```

```python
# Accediendo al ultimo elemento de todas las filas del Array
array_multi[:, -1]
```

## Modificación de un Array

```python
# Creación de un Array unidimensional inicializado con el rango de elementos 0-27
array1 = np.arange(28)  
print("Shape:", array1.shape)   
print("Array 1:", array1)
```

```python
# Cambiar las dimensiones del Array y sus longitudes
array1.shape = (7, 4)
print("Shape:", array1.shape)
array1
```

```python
# El ejemplo anterior devuelve un nuevo Array que apunta a los mismos datos. 
# Importante: Modificaciones en un Array, modificaran el otro Array
array2 = array1.reshape(4, 7)
print("Shape:", array2.shape)
print("Array 2:\n", array2)
```

```python
array1
```

```python
# Modificación del nuevo Array devuelto
array2[0, 3] = 20
print("Array 2:\n", array2)
```

```python
print("Array 1:\n", array1)
```

```python
# No se creo una copia del Array, sino que se creo un nuevo Array que apunta a los mismos datos.
# Es decir, si modificamos el valor de un elemento del Array 1, el valor en el Array 2 tambien cambia.
# No es una copia de datos, sino que es un puntero a los datos.
```

```python
# Desenvuelve el Array, devolviendo un nuevo Array de una sola dimension
# Importante: El nuevo array apunta a los mismos datos
print("Array 1:", array1.ravel()) # puntero a los mismos datos de array1
```

## Operaciones aritméticas con Arrays

```python
# Creación de dos Arrays unidimensionales
array1 = np.arange(2, 18, 2)
array2 = np.arange(8)
print("Array 1:", array1)
print("Array 2:", array2)
```

```python
# Suma
print(array1 + array2) # suma elemento a elemento de los dos Arrays, el resultado es un nuevo Array
```

```python
# Resta
print(array1 - array2) # Resta elemento a elemento de los dos Arrays, el resultado es un nuevo Array
```

```python
# Multiplicacion
# Importante: No es una multiplicación de matrices
print(array1 * array2) # Multiplicación de elemento por elemento de los arrays
```

## Broadcasting

Si se aplican operaciones aritméticas sobre Arrays que no tienen la misma forma (shape) Numpy aplica un propiedad que se denomina Broadcasting.

```python
# Creación de dos Arrays unidimensionales
array1 = np.arange(5)
array2 = np.array([3])
print("Shape Array 1:", array1.shape)
print("Array 1:", array1)
print()
print("Shape Array 2:", array2.shape)
print("Array 2:", array2)
```

```python
# Suma de ambos Arrays
# El broadcasting permite que los Arrays de diferentes dimensiones puedan ser sumados, restados, multiplicados, etc.
# En este caso al array2 se le agregan dimensiones para que pueda ser sumado al array1, el resultado es un nuevo Array
# El array2 quedaria asi: [3, 3, 3, 3, 3]
array1 + array2 
```

```python
# Creación de dos Arrays multidimensional y unidimensional
array1 = np.arange(6)
array1.shape = (2, 3)
array2 = np.arange(6, 18, 4)
print("Shape Array 1:", array1.shape)
print("Array 1:\n", array1)
print()
print("Shape Array 2:", array2.shape)
print("Array 2:", array2)
```

```python
# Suma de ambos Arrays
array1 + array2
```

```python
array1 - array2
```
